-Bagging is an ensemble technique where multiple models are trained independently on different random samples of the dataset and their predictions are combined to produce a final result.  

-Bagging stands for bootstrap aggregating

-Why do we need bagging??  
A single decision tree:  
1)it has high variance   
2)easily overfit   
3)changes alot if data changes slightly   
-Bagging reduces variance and makes predictions more stable.

How Bagging works??
Bootstap sampling:
1)From the original dataset create multiple new datasets
2)Each dataset is created by random sampling with replacement
3)each bootstrap dataset has same size as the original

1)some data points repeat,some datapoints are leftout,those are out of base samples
2)Train multiple models a)Train one model for bootstrap database b)Each tree is trained independently
3)Make Predictions-for regression take average for classification take the majority voting
4)Final Output   a)Aggregrated prediction is more stable and accurate than a single decision tree

When should we use Bagging??
1)where dataset is medium to large
2)model is overfitting
3)decision tree gives unstable result
