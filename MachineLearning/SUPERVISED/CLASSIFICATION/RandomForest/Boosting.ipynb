BOOSTING:
It is an ensemble technique where models are trained sequentially and each new model focus on the mistakes made by previous model 
bagging->paralelly,boosting->sequentially
boosting converts many weak learners into a strong learner

HOW BOOSTING WORKS?
-train first weak model
-identify misclassified and high 
-give more importance to misclassified points
-train next model based on those points
-repeat
-combine all points into a final prediction
