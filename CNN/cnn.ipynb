```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
dataset_path="../CNN/PetImages"
```

```python
datagen= ImageDataGenerator(
    rescale=1./255,


#Data augmentation

shear_range=0.2,
zoom_range=0.2,
horizontal_flip=True,
validation_split=0.2 

)
```

```python
# load training data
# Train data 

train_data = datagen.flow_from_directory(
    dataset_path,
    target_size=(128, 128),
    batch_size=32,
    class_mode='binary',
    subset='training'
)
```

```python
#Load validation data

val_data=datagen.flow_from_directory(
    dataset_path,
    target_size=(128,128),
    batch_size=32,
    class_mode='binary',
    subset='validation'
)
```

```python
print(train_data.class_indices)
```

```python
#Building CNN model

model=keras.Sequential()
```

```python
#First Convulational layer
#Based on the dataset and its complexity we decide how many convulational and pooling layers are required.


model.add(
    keras.layers.Conv2D(
        filters=32,
        kernel_size=(3,3),
        activation='relu',
        input_shape=(128,128,3),
    )
)
```

```python
#First pooling layer

model.add(

    keras.layers.MaxPooling2D(
        pool_size=(2,2)
    ),
)
```

```python
# 2 Conv Layer

model.add(
    keras.layers.Conv2D(
        filters=32,
        kernel_size=(3,3),
        activation='relu'
    )
)
```

```python
# 2 Pooling Layer

model.add(

    keras.layers.MaxPooling2D(
        pool_size=(2,2)
    )
)
```

```python
# Flatten

model.add(keras.layers.Flatten())
```

```python
# Dense Layer

model.add(
    keras.layers.Dense(
        units=128,
        activation='relu'
    )
)
```

```python
# Output Layer

model.add(
    keras.layers.Dense(
        units=1,
        activation='sigmoid'
    )
)
```

```python
# Compile CNN
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```
