```python
# eating,eat,eaten--> eat
# going,gone,goes --> go
```

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
# PorterStemmer

from nltk.stem import PorterStemmer
stemming=PorterStemmer()
```

```python
for word in words:
    print(word+"------>"+stemming.stem(word))
```

```python
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
```

```python
sentences=sent_tokenize(sentences)
```

```python
# Apply stopwords and filter and then apply stemming

for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[stemmer.stem(word) for  word in words if word not in set(stopwords.words("english"))]
    sentences[i]=' '.join(words)
```

```python
from nltk.stem import WordNetLemmatizer
lemmitizer=WordNetLemmatizer()
```

```python
for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[lemmitizer.lemmatize(word.lower(),pos='v') for word in words if word not in set(stopwords.words('english'))]
    sentences[i]=' '.join(words)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

text = """
Yesterday I decided to become a serious data scientist and start my day with productivity.
I opened my laptop, created a new Python file, and confidently wrote import pandas as pd.
After that I stared at the screen like the code would magically write itself.
To feel smarter, I opened five Stack Overflow tabs and two YouTube tutorials.
One tutorial said machine learning is easy, but my code disagreed with 37 error messages.
Then my friend asked what I was doing, and I proudly said I was training a model.
In reality, the only thing getting trained was my patience.
My laptop fan started sounding like a rocket preparing for launch.
Meanwhile my coffee became cold because debugging is a full-time job.
I fixed one bug and suddenly three new bugs appeared like a software family reunion.
At some point I wasn’t sure if I was coding or just negotiating with Python.
Even my cat looked at the screen like it understood the problem better than me.
Finally the program ran successfully and printed Hello World.
I felt like I had just built the world’s most advanced artificial intelligence.
Then I realized the real model being trained was actually me.
"""

# Initialize CountVectorizer
vectorizer = CountVectorizer()

# Fit and transform the text
X = vectorizer.fit_transform([text])

# Convert to DataFrame for readability
bow_df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())

print(bow_df)
```
